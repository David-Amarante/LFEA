# 2. Study of Detector Efficiency

**Objective:** Measure detection efficiency for β and γ radiation.

**Source Information:**
- $^{204}$Tl (Thallium-204): Activity 0.5 µCi in April 2008
  - Manufacturer uncertainty: ±20% on activity
  - Emits β particles (main) and some γ
  - Metal disc: Ø 7.25 ± 0.05 mm (defines geometric efficiency)

- $^{137}$Cs (Cesium-137): Activity 5 µCi in March 1994, source L02 (thin window)
  - Manufacturer uncertainty: ±20% on activity
  - Emits γ rays (main transition: 661.7 keV)
  - Measure only γ radiation (block β)

**Efficiency Definition:** ε = N_det / (N_source × G)
- N_det: detected counts
- N_source: source activity × time
- G: geometric efficiency factor

## Imports and Constants

```python
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime

# Physical constants
HALF_LIFE_TL204 = 3.78 * 365.25 * 24 * 3600  # seconds (3.78 years)
HALF_LIFE_CS137 = 30 * 365.25 * 24 * 3600  # seconds (30 years)

# Activity decay: A(t) = A0 * exp(-λ*t)
# λ = ln(2) / T_half
```

## Activity Decay Calculation

```python
def decay_activity(a0, t_half, elapsed_time):
    """Calculate activity after decay.
    
    Args:
        a0: Initial activity (Bq or µCi)
        t_half: Half-life in same units as elapsed_time
        elapsed_time: Time elapsed in same units
        
    Returns:
        a: Current activity
    """
    lambda_decay = np.log(2) / t_half
    return a0 * np.exp(-lambda_decay * elapsed_time)

# Current date (analysis date)
date_analysis = datetime(2026, 5, 15)

# ===== Tl-204 Decay =====
# Original: 0.5 µCi in April 2008 (middle of month ± 15 days)
tl204_initial = 0.5  # µCi
tl204_initial_unc = 0.5 * 0.20  # 20% manufacturer uncertainty
date_measurement_tl204 = datetime(2008, 4, 15)  # Middle of April
date_measurement_tl204_min = datetime(2008, 3, 31)  # 15 days before
date_measurement_tl204_max = datetime(2008, 4, 30)  # 15 days after

elapsed_tl204_nominal = (date_analysis - date_measurement_tl204).total_seconds()
elapsed_tl204_min = (date_analysis - date_measurement_tl204_max).total_seconds()  # shortest decay
elapsed_tl204_max = (date_analysis - date_measurement_tl204_min).total_seconds()  # longest decay

tl204_current = decay_activity(tl204_initial, HALF_LIFE_TL204, elapsed_tl204_nominal)
tl204_current_min = decay_activity(tl204_initial - tl204_initial_unc, HALF_LIFE_TL204, elapsed_tl204_min)
tl204_current_max = decay_activity(tl204_initial + tl204_initial_unc, HALF_LIFE_TL204, elapsed_tl204_max)

print("="*70)
print("Tl-204 (β source)")
print("="*70)
print(f"Original activity: {tl204_initial} µCi (April 15, 2008)")
print(f"  Activity uncertainty: ±{tl204_initial_unc:.2f} µCi (±20% manufacturer)")
print(f"  Time uncertainty: April 1 to 30, 2008 (±15 days)")
print(f"\nCurrent activity (May 15, 2026):")
print(f"  Nominal: {tl204_current:.6f} µCi")
print(f"  Range: {tl204_current_min:.6f} to {tl204_current_max:.6f} µCi")
print(f"  Uncertainty: ±{(tl204_current_max - tl204_current_min)/2:.6f} µCi")

# ===== Cs-137 Decay =====
# Original: 5.0 µCi in March 1994 (middle of month ± 15 days)
cs137_initial = 5.0  # µCi
cs137_initial_unc = 5.0 * 0.20  # 20% manufacturer uncertainty
date_measurement_cs137 = datetime(1994, 3, 15)  # Middle of March
date_measurement_cs137_min = datetime(1994, 2, 28)  # 15 days before
date_measurement_cs137_max = datetime(1994, 3, 31)  # 15 days after

elapsed_cs137_nominal = (date_analysis - date_measurement_cs137).total_seconds()
elapsed_cs137_min = (date_analysis - date_measurement_cs137_max).total_seconds()  # shortest decay
elapsed_cs137_max = (date_analysis - date_measurement_cs137_min).total_seconds()  # longest decay

cs137_current = decay_activity(cs137_initial, HALF_LIFE_CS137, elapsed_cs137_nominal)
cs137_current_min = decay_activity(cs137_initial - cs137_initial_unc, HALF_LIFE_CS137, elapsed_cs137_min)
cs137_current_max = decay_activity(cs137_initial + cs137_initial_unc, HALF_LIFE_CS137, elapsed_cs137_max)

print("\n" + "="*70)
print("Cs-137 (γ source, L02 thin window)")
print("="*70)
print(f"Original activity: {cs137_initial} µCi (March 15, 1994)")
print(f"  Activity uncertainty: ±{cs137_initial_unc:.2f} µCi (±20% manufacturer)")
print(f"  Time uncertainty: March 1-31, 1994 (±15 days)")
print(f"\nCurrent activity (May 15, 2026):")
print(f"  Nominal: {cs137_current:.6f} µCi")
print(f"  Range: {cs137_current_min:.6f} to {cs137_current_max:.6f} µCi")
print(f"  Uncertainty: ±{(cs137_current_max - cs137_current_min)/2:.6f} µCi")
```

```python
from pathlib import Path
import re

def load_counting_stats_file_local(file_path):
    text = Path(file_path).read_text(encoding='utf-8', errors='ignore').splitlines()
    avg = None
    std = None
    counts = []
    for i, line in enumerate(text):
        if 'Average Count' in line:
            m = re.search(r'=\s*(\d+)', line)
            if m:
                avg = float(m.group(1))
        if 'Standard Deviation' in line:
            m = re.search(r'=\s*(\d+)', line)
            if m:
                std = float(m.group(1))
        match = re.match(r'\s*\d+,\s*(\d+)\s*', line)
        if match and i > 5:
            counts.append(float(match.group(1)))
    return avg, std, np.array(counts)

# Load background and beta stats (Tl-204)
# Notebook lives in the geiger-muller folder, so use current directory
data_dir = Path('.')
bg_file = data_dir / 'bstatics.asc'
beta_file = data_dir / 'betastat.asc'

if bg_file.exists():
    bg_avg, bg_std, bg_counts = load_counting_stats_file_local(bg_file)
else:
    bg_avg = bg_std = 0

if beta_file.exists():
    beta_avg, beta_std, beta_counts = load_counting_stats_file_local(beta_file)
    beta_net = beta_avg - bg_avg
    beta_net_unc = np.sqrt(beta_std**2 + bg_std**2)
    print(f"Loaded betastat: raw {beta_avg:.1f} ± {beta_std:.1f} → net {beta_net:.1f} ± {beta_net_unc:.1f} counts")
else:
    print('betastat.asc not found in current directory')
```

```python
# Load Cs-137 measurements (gamma, blocked beta) and compute efficiency
cal_file = Path('calstat.asc')
cd_file = Path('cnadastat.asc')
acquisition_time = 10.0  # seconds per counting acquisition (assumed)

for fpath, name in [(cal_file, 'calstat.asc'), (cd_file, 'cnadastat.asc')]:
    if fpath.exists():
        avg, std, counts = load_counting_stats_file_local(fpath)
        net = avg - bg_avg
        net_unc = np.sqrt(std**2 + bg_std**2)
        # Convert Cs-137 activity (µCi -> Bq)
        activity_bq = cs137_current * 3.7e4  # 1 µCi = 3.7e4 Bq
        # Use geometric efficiency placeholder G=1 (apparent efficiency)
        G = 1.0
        eff, eff_unc = calculate_efficiency(n_detected=net, n_background=0, activity_bq=activity_bq, time_s=acquisition_time, geometric_eff=G)
        print(f"\n{name}: raw {avg:.1f} ± {std:.1f} -> net {net:.1f} ± {net_unc:.1f} counts")
        print(f"  Activity (Cs-137) = {cs137_current:.6f} µCi = {activity_bq:.2f} Bq")
        print(f"  Apparent efficiency (G=1): ε = {eff:.6e} ± {eff_unc:.6e} (fraction)")
    else:
        print(f"{name} not found")
```

```python
# Compute intrinsic efficiency for Cs-137 using activity and real counts
# Assumptions: acquisition_time (s) and geometric efficiency G provided below
acquisition_time = 10.0  # seconds (adjust if different)
G = 1.0  # geometric factor (set to 1 for detector-face measurements)

# Use calstat if available
cal_file = Path('calstat.asc')
if cal_file.exists():
    cal_avg, cal_std, cal_counts = load_counting_stats_file_local(cal_file)
    cal_net = cal_avg - bg_avg
    cal_net_unc = np.sqrt(cal_std**2 + bg_std**2)
    # activity in Bq
    activity_bq = cs137_current * 3.7e4  # µCi -> Bq
    # activity uncertainty (20% relative as provided)
    activity_unc = 0.20 * activity_bq
    # Efficiency (fraction)
    eff = cal_net / (activity_bq * acquisition_time * G)
    # Uncertainty propagation: counts uncertainty + activity uncertainty
    sigma_eff_counts = cal_net_unc / (activity_bq * acquisition_time * G)
    sigma_eff_activity = (cal_net * activity_unc) / (activity_bq**2 * acquisition_time * G)
    sigma_eff = np.sqrt(sigma_eff_counts**2 + sigma_eff_activity**2)
    print("\nComputed intrinsic efficiency for Cs-137 (using calstat):")
    print(f"  Net counts = {cal_net:.1f} ± {cal_net_unc:.1f} counts (per acquisition)")
    print(f"  Activity = {activity_bq:.2f} Bq (±{activity_unc:.2f} Bq, 20%)")
    print(f"  Acquisition time = {acquisition_time:.1f} s")
    print(f"  Geometric factor G = {G}")
    print(f"\n  Efficiency ε = {eff:.6e} ± {sigma_eff:.6e} (fraction)")
else:
    print('calstat.asc not found; cannot compute Cs-137 efficiency')
```

## Efficiency Calculation and Comparison

```python
# Efficiency formula: ε = N_det / (A × t × G)
# where:
#   N_det = detected counts (background-corrected)
#   A = source activity (Bq or decays/s)
#   t = acquisition time (s)
#   G = geometric efficiency

def calculate_efficiency(n_detected, n_background, activity_bq, time_s, geometric_eff):
    """Calculate detection efficiency.
    
    Args:
        n_detected: Total detected counts
        n_background: Background counts
        activity_bq: Source activity in Bq
        time_s: Acquisition time in seconds
        geometric_eff: Geometric efficiency factor (0-1)
        
    Returns:
        efficiency: Detection efficiency (fraction)
        uncertainty: Uncertainty on efficiency
    """
    n_net = n_detected - n_background
    denominator = activity_bq * time_s * geometric_eff
    
    if denominator <= 0:
        return np.nan, np.nan
        
    efficiency = n_net / denominator
    
    # Uncertainty from counting statistics
    var_net = n_detected + n_background  # Poisson
    var_efficiency = (var_net / denominator**2)
    uncertainty = np.sqrt(var_efficiency)
    
    return efficiency, uncertainty

print("Efficiency calculation functions defined")
```
