# Plot Simples - Canais vs Counts

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import re
```

```python
def plot_asc(filename):
    """Lê arquivo .ASC e faz plot simples de canais vs counts"""
    text = Path(filename).read_text(encoding='utf-8', errors='ignore').splitlines()
    channels = []
    counts = []
    
    for line in text:
        match = re.match(r'\s*(\d+),\s*(\d+)', line)
        if match:
            channels.append(int(match.group(1)))
            counts.append(int(match.group(2)))
    
    channels = np.array(channels)
    counts = np.array(counts)
    
    fig, ax = plt.subplots(figsize=(12, 6))
    ax.plot(channels, counts, 'b-', linewidth=1.5)
    ax.set_yscale('log')
    ax.set_xlabel('Channel')
    ax.set_ylabel('Counts')
    ax.set_title(filename)
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```

## Plot

```python
plot_asc('p02.ASC')
plot_asc('p04.ASC')
plot_asc('p06.ASC')
plot_asc('p08.ASC')
plot_asc('p10.ASC')
plot_asc('p12.ASC')
```
