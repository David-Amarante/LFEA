# 4. Study of Attenuation of Particles in Matter

**Objective:** Study how radiation intensity decreases when passing through matter.

**Radiation types:**
- β particles (from $^{204}$Tl)
- γ rays (from $^{137}$Cs)

**Materials:** Different absorber materials with varying mass thickness (low and high)

**Expected behavior:**
- **β attenuation:** I = I₀ × exp(-x/R_e)
  - Continuous spectrum, finite range R_e
  - Bragg-Kleeman rule for range

- **γ attenuation:** I = I₀ × exp(-μx)
  - Linear attenuation coefficient μ (cm⁻¹) or μ/ρ (cm²/g)
  - Compton scattering, pair production, photoelectric effect

## Imports and Setup

## Attenuation Models

```python
def gamma_attenuation(x, i0, mu, background):
    """Exponential attenuation for gamma rays.
    
    Args:
        x: Thickness (cm)
        i0: Initial intensity
        mu: Linear attenuation coefficient (cm⁻¹)
        background: Background count rate
        
    Returns:
        I: Transmitted intensity
    """
    return i0 * np.exp(-mu * x) + background

def beta_attenuation_bragg_kleeman(x, i0, r_e, background):
    """β attenuation with Bragg-Kleeman rule.
    
    For practical use in this experiment,
    assume linear initial drop then longer tail.
    
    Args:
        x: Thickness (mg/cm²)
        i0: Initial intensity
        r_e: Effective range (mg/cm²)
        background: Background
        
    Returns:
        I: Transmitted intensity
    """
    # Simplified model: linear drop to range, then tail
    attenuation = np.maximum(1 - x / r_e, 0.1)  # Some tail remains
    return i0 * attenuation + background

print("Attenuation models defined")
```
