# Geiger-Müller Detector Experiment

## Introduction

This series of experiments studies the characteristics of the Geiger-Müller detector and applications of particle counting.

**Objectives:**
1. Study detector characteristics: detection principle, operating mode, particle types, count rate, detection efficiency
2. Study applications with particle counts: environmental radioactivity, electron backscatter, attenuation in matter, statistics

**Source:** $^{204}$Tl (Thallium-204) - 0.5 µCi, emits β particles and γ rays
