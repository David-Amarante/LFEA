# 3. Study of Intensity Variation with Distance

**Objective:** Measure how count rate varies with distance from the source.

**Source:** $^{137}$Cs (661.7 keV γ rays)

**Method:** Use shelves 1-5 at defined distances to measure count rate vs. distance.

**Expected behavior:** I ∝ 1/r² (inverse square law for point source)

Observed rate: N(r) = N₀/r² + B
- N₀: normalization constant
- B: background rate
- r: distance from source

## Imports and Setup

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

print("Distance variation study initialized")
```

## Inverse Square Law Model

```python
def inverse_square_law(r, n0, background):
    """Inverse square law with background.
    
    Args:
        r: Distance (m)
        n0: Normalization constant (counts/s × m²)
        background: Background count rate (counts/s)
        
    Returns:
        N: Count rate (counts/s)
    """
    return n0 / (r**2) + background

print("Inverse square law model defined")
```
